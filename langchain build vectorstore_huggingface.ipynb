# Build an efficient retrieval system

In this script, we are processing a collection of text documents related to Nordic travel literature. The script is structured into three key steps: loading, processing, and querying. The overall goal is to extract relevant information from the documents.

## Loading 

The script begins by importing necessary libraries and suppressing any warnings that may arise during execution. This is done using the warnings module, which helps to keep the output clean and focused on the results of interest.

Next, we change the working directory to the folder containing the Nordic Exceptionalism Corpus. This is accomplished using the os module, which allows us to navigate the file system programmatically.

We then create a list of all filenames in the directory and load the text files using `UnstructuredFileLoader` from the LangChain library. Each file is processed, and the contents are stored in a list called `docs`.

To facilitate easier processing and analysis, the script splits the loaded documents into smaller chunks. This is done using the `RecursiveCharacterTextSplitter`, which allows for a specified chunk size (1500 characters) and overlap (150 characters) between chunks. The chosen parameters ensure that the text is divided meaningfully while maintaining context.

### Code: Load Documents and Split into Chunks

This code cell imports the necessary libraries, navigates to the document directory, loads all text files using `UnstructuredFileLoader`, and splits them into chunks using `RecursiveCharacterTextSplitter` with a chunk size of 1500 characters and an overlap of 150 characters.


```python
import warnings
warnings.filterwarnings('ignore')
from langchain.document_loaders import UnstructuredFileLoader
import os

# Define the directory containing the text files (relative path)
doc_directory = os.path.abspath(r'./literature txt')

# Get list of files in the directory
files = os.listdir(doc_directory)

# Create full paths for each file
file_paths = [os.path.join(doc_directory, file) for file in files]

# Load txt files using full paths
loaders = [UnstructuredFileLoader(file_path) for file_path in file_paths]
docs = []
for loader in loaders:
    docs.extend(loader.load())

# Split documents into chunks
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1500,
    chunk_overlap=150,
    separators=["\n\n", "\n", "(?<=\. )", " ", ""]
)
splits = text_splitter.split_documents(docs)
```

### Processing

This section covers the embedding generation and vector store creation process.

**Embedding Generation**: The next step involves generating embeddings for the document chunks. Embeddings are numerical representations of text that capture semantic meaning, allowing for more effective searching and retrieval. The script utilizes the HuggingFaceEmbeddings model, specifically the `sentence-transformers/all-MiniLM-L6-v2` model, to create these embeddings.

**Vector Store Creation**: With the embeddings generated, the script then creates a Chroma vector store. This store allows for efficient querying of the document chunks based on their embeddings. The vector store is set to persist in a `chroma/` directory in the same location as this notebook, ensuring that the embeddings and document relationships are saved for future use.

### Code: Generate Embeddings and Create Vector Store

This code cell initializes the HuggingFace embedding model and creates a persistent Chroma vector store from the document splits. The vector store will be saved to the specified directory for future use.


```python
import os
# Embeddings
from langchain.embeddings import HuggingFaceEmbeddings

# Initialiser embedding modellen
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Generate embeddings for the documents
# Build a Chroma vectorstore
from langchain.vectorstores import Chroma

# Vector store will be saved in the same directory as this notebook
persist_directory = 'chroma/'

vectordb = Chroma.from_documents(
    documents=splits,
    embedding=embedding_model,
    persist_directory=persist_directory
)
```

### Querying using the max_marginal_relevance_search method

This section demonstrates how to query the vector store using the `max_marginal_relevance_search` method. This approach strives to achieve both relevance to the query and diversity among the results, which helps avoid retrieving very similar documents.

The script poses a specific question to the vector store: "find examples of people in Scandinavia (Denmark, Norway, Sweden, Iceland) being trustworthy." The method retrieves `k=3` results from `fetch_k=10` candidates, ensuring a balance between relevance and diversity.

Finally, the script iterates through the retrieved documents, formatting and printing the results. Each example is displayed along with its source, providing context for the information presented. The use of string manipulation ensures that the output is clean and readable.

```python
# Send question to the vectorstore
question = 'find examples of people in Scandinavia (Denmark, Norway, Sweden, Iceland) being trustworthy.'
docs_mmr = vectordb.max_marginal_relevance_search(question,k=3, fetch_k=10)

import re
count = 0
for i in docs_mmr:
    count = count + 1
    print ('Example: '+ str(count))
    print ('\n')
    text_out = str(i.page_content).replace('\n', ' ') 
    print (text_out)
    print ('\n')
    source = str(i.metadata).replace('}','').replace('.txt','').split('\\')[-1]
    print (f"Source: {source}")
    print ('*'*50)
    print ('\n')


```

### Querying using the similarity_search method

This section demonstrates how to query the vector store using the `similarity_search` method. This approach retrieves the most semantically similar document chunks based on cosine similarity of embeddings.

The script poses the same question to the vector store: "find examples of people in Scandinavia (Denmark, Norway, Sweden, Iceland) being trustworthy." Using the `similarity_search` method, it retrieves the top `k=5` relevant document chunks that address this query.

Finally, the script iterates through the retrieved documents, formatting and printing the results. Each example is displayed along with its source, providing context for the information presented. The use of string manipulation ensures that the output is clean and readable.

```python
question = 'find examples of people in Scandinavia (Denmark, Norway, Sweden, Iceland) being trustworthy.'
docs_ss = vectordb.similarity_search(question, k=5)

import re
count = 0
for i in docs_ss :
    count = count + 1
    print ('Example: '+ str(count))
    print ('\n')
    text_out = str(i.page_content).replace('\n', ' ') 
    print (text_out)
    print ('\n')
    source = str(i.metadata).replace('}','').replace('.txt','').split('\\')[-1]
    print (f"Source: {source}")
    print ('*'*50)
    print ('\n')
```

## Loading existing vector stores

If you have already built a vector store and saved it, you can load it directly without having to build it again. This is much faster than rebuilding the entire vector store from scratch.

```python
# Load existing vector stores
import os
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import Chroma

# Use the same embedding model as when you built the vector store
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# The path to the saved vector store (same directory as this notebook)
persist_directory = 'chroma/'

# Load the existing vector store
# Use Chroma() instead of Chroma.from_documents() when loading an existing
vectordb = Chroma(
    persist_directory=persist_directory,
    embedding_function=embedding_model
)

print("Vector store loaded!")
print(f"Number of documents in the vector store:{vectordb._collection.count()}")
```

### Example Query: Cultural Descriptions

This example demonstrates querying the loaded vector store for cultural descriptions of people and inhabitants. The query uses the `similarity_search` method to find the top 3 most relevant document chunks.


```python
# Example query after loading existing vector stores
# You can now use all the same query methods as before

question = 'find examples with cultural descriptions of people and inhabitants'
docs_ss = vectordb.similarity_search(question, k=3)

import re
count = 0
for i in docs_ss:
    count = count + 1
    print('Example: ' + str(count))
    print('\n')
    text_out = str(i.page_content).replace('\n', ' ') 
    print(text_out)
    print('\n')
    source = str(i.metadata).replace('}','').replace('.txt','').split('\\')[-1]
    print(f"Source: {source}")
    print('*'*50)
    print('\n')
```
